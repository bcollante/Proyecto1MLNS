```python
import kagglehub
import pandas as pd
from kagglehub import KaggleDatasetAdapter
from pathlib import Path
# Descarga la última versión del dataset (ejemplo: Titanic)
DATASET = "steubk/wikiart"
path=kagglehub.dataset_download(DATASET, path='Abstract_Expressionism/aaron-siskind_acolman-1-1955.jpg')

path
```

```python
df = pd.read_csv("classes.csv")
print(df[["filename", "genre"]].head())
```

```python
styles = [
    "Impressionism",
    "Cubism",
    "Baroque",
    "Ukiyo e",
    "Romanticism",
    "Action painting"
]
import ast

df["style"] = df["genre"].apply(
    lambda x: ast.literal_eval(x)[0]
)
```

```python
selected = (
    df[df["style"].isin(styles)]
    .groupby("style", group_keys=False)
    .sample(n=10, random_state=42)
)
print(selected[["filename", "style"]])
```

```python
for filename in selected["filename"]:

    path = kagglehub.dataset_download(
        DATASET,
        path=filename
    )

    print(path)
```

```python
import numpy as np
import os
import cv2
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans, MeanShift, estimate_bandwidth
from PIL import Image
```

```python
print(selected["style"].unique())
```

```python
style = "Cubism"
imagenes = selected[selected["style"] == style]
fig, axes = plt.subplots(2, 5, figsize=(15, 7))

for ax, (_, row) in zip(axes.flatten(), imagenes.iterrows()):

    path = kagglehub.dataset_download(
        DATASET,
        path=row["filename"]
    )

    img = Image.open(path)

    ax.imshow(img)
    ax.axis("off")

plt.suptitle(style, fontsize=18)
plt.tight_layout()
plt.show()
```
